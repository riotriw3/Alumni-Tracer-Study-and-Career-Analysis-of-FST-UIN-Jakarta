WHO COVID-19 Global Data memberikan informasi komprehensif mengenai pandemi global COVID-19 seperti yang dilaporkan ke Organisasi Kesehatan Dunia (WHO). Ukuran data yang digunakan adalah 2.294 kB dengan jumlah pengukuran data adalah 49680 baris dan 8 kolom. Untuk tipe-tipe data yang digunakan dapat dilihat pada output yang sudah dirunning.
Berikut Penjelasan dari WHO COVID-19 Global Data yang akan digunakan:

1.   Date_reported (Date): Tanggal pelaporan ke WHO.
2.   Country_code (String): Kode negara ISO Alpha-2.
3.   Country (String): Nama negara, wilayah, atau wilayah.
4.   WHO_region (String): WHO regional offices (Kantor regional WHO). Negara-negara Anggota WHO dikelompokkan menjadi enam wilayah WHO, yaitu AFRO (Kantor Regional untuk Afrika), AMRO (Kantor Regional untuk Amerika), SEARO (Kantor Regional untuk Asia Tenggara), EURO (Kantor Regional untuk Eropa), EMRO (Kantor Regional untuk Mediterania Timur), dan WPRO (Kantor Regional untuk Pasifik Barat).
5.   New_cases (Integer): Jumlah kasus terkonfirmasi baru yang dilaporkan pada hari tertentu. Hal ini dihitung dengan mengurangkan jumlah kasus kumulatif sebelumnya dari jumlah kasus kumulatif saat ini
6.   Cumulative_cases (Integer): Total kumulatif kasus terkonfirmasi yang dilaporkan ke WHO hingga tanggal yang ditentukan.
7.   New_deaths (Integer): Jumlah kematian baru yang dikonfirmasi yang dilaporkan pada hari tertentu. Mirip dengan kasus baru, penghitungan ini dilakukan dengan mengurangkan jumlah kematian kumulatif sebelumnya dari jumlah kematian kumulatif saat ini
8.   Cumulative_deaths (Integer): Total kumulatif kematian terkonfirmasi yang dilaporkan ke WHO hingga tanggal yang ditentukan.tem daftar

Vaccination Data memberikan informasi komprehensif mengenai negara-negara yang memberikan vaksinasi kepada warganya. Kumpulan data ini disediakan oleh WHO (berdasarkan negara). Ukuran data yang digunakan adalah 21 kB dengan jumlah pengukuran data adalah 210 baris dan 17 kolom. Untuk tipe-tipe data yang digunakan dapat dilihat pada output yang sudah dirunning.
Berikut Penjelasan dari WHO COVID-19 Global Data yang akan digunakan:

Columns Description:
1. COUNTRY (Object): Nama negara
2. ISO3 (Object): Kode negara ISO Alpha-3
3. WHO_REGION (Object): WHO regional offices (Kantor regional WHO). Negara-negara Anggota WHO dikelompokkan menjadi enam wilayah sebagai berikut, AFRO (Kantor Regional untuk Afrika), AMRO (Kantor Regional untuk Amerika), SEARO (Kantor Regional untuk Asia Tenggara), EURO (Kantor Regional untuk Eropa), EMRO (Kantor Regional untuk Mediterania Timur), dan WPRO (Kantor Regional untuk Pasifik Barat).
4. DATA_SOURCE (Object): Dua sumber data, reporting: Data yang dilaporkan oleh Negara Anggota, atau bersumber dari laporan resmi, dan OWID
5. DATE_UPDATED (Object): Tanggal pembaruan terakhir
6. TOTAL_VACCINATIONS (Object): Total kumulatif dosis vaksin yang diberikan
7. PERSONS_VACCINATED_1PLUS_DOSE (Float64): Jumlah kumulatif orang yang menerima setidaknya satu dosis vaksin
8. TOTAL_VACCINATIONS_PER100 (Float64): Total dosis vaksin kumulatif yang diberikan per 100 penduduk
9. PERSONS_VACCINATED_1PLUS_DOSE_PER100(Float64) Kumulatif orang yang divaksinasi dengan setidaknya satu dosis per 100 penduduk
10. PERSONS_LAST_DOSE (Float64): dosis terakhir orang
11. PERSONS_LAST_DOSE_PER100 (Float64): dosis terakhir orang per 100 penduduk
12. VACCINES_USED (Float64): Gabungan nama pendek vaksin: “Perusahaan - Nama Produk”
13. FIRST_VACCINE_DATE (Object): Tanggal vaksinasi pertama. Setara dengan tanggal mulai/peluncuran vaksin pertama yang diberikan di suatu negara
14. NUMBER_VACCINES_TYPES_USED (Object): Jumlah jenis vaksin yang digunakan per negara, wilayah, wilayah
15. PERSONS_BOOSTER_ADD_DOSE (Float64): orang yang tambah dosis booster
16. PERSONS_BOOSTER_ADD_DOSE_PER100 (Float64): orang yang tambah dosis booster per 100 penduduk
17. Unnamed (Float64): kolom tanpa nama

```python
#Menggabungkan data urut tanggal terbaru dengan data vaksin berdasarkan country,
#karena date reported di vaksin tidak stiap saat
whocov_vac_merge <- merge(whocov_urut_newest, vac, by = c("Country", "WHO_region"))
#Membuat datset jumlah negara di tiap who region
WHO_sum <- whocov_vac_merge %>%
  group_by(WHO_region) %>%
  summarise(Total_country = size_sum(WHO_region))
WHO_sum$Total_country <- gsub("\\[|\\]", "", WHO_sum$Total_country)
WHO_sum$Total_country <- as.integer(WHO_sum$Total_country)
```

Menjumlahkan kasus COVID-19 menjadi menjadi periode mingguan

```python
#* Time Series Plot Mingguan
#Membuat kolom yang membagi tanggal menjadi mingguan dan tahunan
whocov <- whocov %>%
  mutate(Week = lubridate::week(Date_reported),
         Year = lubridate::year(Date_reported))
```

```python
#Menjumlahkan kasus mingguan di tiap tahunnya
summarized_COVID <- whocov %>%
  group_by(WHO_region, Week, Year) %>%
  summarise(Total_New_Cases = sum(New_cases),
            Total_New_Deaths = sum(New_deaths), .groups = 'drop')
```

```python
#timeseries kasus mingguan
ggplot(summarized_COVID) +
  geom_col(aes(x = Week, y = Total_New_Cases/1000000, fill = WHO_region)) +
  facet_grid(~ Year) +
  labs(title = "Total New Cases per Week in 2020-2023",
       x = "Week",
       y = "Total New Cases (dalam jutaan)") +
  theme_minimal()
```
